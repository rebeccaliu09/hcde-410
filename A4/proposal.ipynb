## 1. Overview & Research Questions

When I watch Formula 1, team radio is one of the few ways I get to hear what drivers and their engineers are saying during a race. The clips can make a driver come across as calm, frustrated, or funny. I’m curious about something that comes before those impressions: whose voice appears most often in the clips we can access, and when do we hear it?

I will use the team radio clips available through OpenF1 for the 2023 and 2024 race sessions. This is an exploratory study of who appears in those clips and how the timing varies across races.

1. **Who.** Which drivers appear most often in the available clips, and how does the ranking change when I account for how many races each driver started?
2. **When.** What share of each race’s available in-race clips falls in its early, middle, and late stages, and how does that pattern vary across races?

## 2. Background / Related Work

Shoemaker and Vos (2009) describe *gatekeeping* as the process through which media organizations decide what information reaches audiences. I’m applying that idea to team radio. How often we hear a driver depends in part on which clips are made available in the first place.

Collantine (2025) interviewed F1’s director of broadcast and media about how the production team decides which radio exchanges to play on the global television feed. The interview explains how editors look for clips that help tell the story unfolding on track. OpenF1’s team radio documentation also describes its dataset as a limited selection of communications. Together, these sources help define what I can measure. I can compare how often drivers appear in the OpenF1 collection. Estimating each driver’s chance of being selected would require a record of all their radio exchanges, and describing the global TV broadcast would require checking which clips actually aired there.

## 4. Methodology

**Prepare and check the data.** I will check for duplicate clips and missing driver IDs or timestamps and report how many rows are excluded from each analysis. I will also summarize clip counts by race so readers can see how coverage varies.

**RQ1 · Driver representation.** I will use race results to identify who actually started each race in the dataset, then match those starts to the radio records by race and driver. I will count available clips by driver and divide each count by the driver’s number of starts in those races. Starts with zero available clips will count in the denominator. This lets me compare drivers who raced different numbers of times. I will make a horizontal bar chart ranked by clips per start and report each driver’s total clips and starts alongside it so readers can see when an average is based on only a few races.

**RQ2 · Timing within a race.** I will check the start and finish times against race timing records, then split each race’s elapsed time into three equal parts. I will calculate the share of a race’s clips in each part using clips recorded between the start and finish. Using thirds lets me compare races with different durations. I will report any races I cannot place on a reliable timeline. The figure will use one 100% stacked horizontal bar per race, grouped by season, with the number of in-race clips shown alongside each bar.

**Small audio sample.** I tested a few audio links and confirmed that they played. After the timing analysis, I will select two races with playable recordings where radio clips appear at different stages of the race. I will use Whisper to transcribe a small sample from each race, listen back to check the transcripts, and flag any unclear speech. I will use a few clear examples to illustrate what drivers and their engineers talked about, such as strategy, car issues, or reactions.

## 5. Limitations, Ethics & Contingencies

- **Coverage.** The dataset represents clips available through OpenF1. A driver’s clip count cannot tell me how often that driver communicated with the team overall.
- **Audio availability and interpretation.** The audio links may stop working. The two main analyses can proceed with the saved structured data. For the audio examples, I will link to source clips where available and give enough race context to make the quotes understandable. Noise or unclear speech will be marked as uncertain.
- **Presentation.** The main findings will be presented with static charts. An interactive chart with audio links is a stretch goal.
